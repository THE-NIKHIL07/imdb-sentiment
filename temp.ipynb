```python
from tensorflow.keras.preprocessing.text import one_hot

```

```python
#senetences
sent=[
    'I love my dog',
    'I love my cat',    
    'You love my dog!',
    'Do you think my dog is amazing?',
    'I think my dog is amazing!',
    'your videos are amazing',
    'understand the concept of RNN',
    'RNN is a type of neural network',
 ]
```

```python
#define the vocabulary size
voc_size=10000
#one hot representation
onehot_repr=[one_hot(words,voc_size)for words in sent]
print(onehot_repr)#word representation as integers indexes
```

```python
#word embedding
from tensorflow.keras.layers import Embedding
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
import numpy as np
sent_length=8
embedded_docs=pad_sequences(onehot_repr,padding='pre',maxlen=sent_length)
```

```python
print(embedded_docs)#padding the documents to a fixed length
```

```python
#feature dimension
dim=10
model=Sequential()
model.add(Embedding(voc_size,output_dim=dim,input_length=sent_length))
model.compile('adam','mse')
```

```python
model.summary()
```

```python
model.predict(embedded_docs)
```
